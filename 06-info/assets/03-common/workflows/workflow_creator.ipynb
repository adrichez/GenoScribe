### **Workflow completo en color**

```python
# Configuración general
input_path = "output/general/general_workflow.png"

width = 1275  # misma base para todos los recuadros
height = 1277  # misma altura para todos los recuadros
cy = 5569  # misma coordenada Y céntrica para todos los recuadros 

color = "#272EF5"
grosor = 15

# Lista con: (output_path, cx, cy)
analisis = [
    ("output/specific/transcriptomics_bulk_rna_seq_workflow.png", 710, cy),
    ("output/specific/transcriptomics_sc_rna_seq_workflow.png", 2190, cy),
    ("output/specific/transcriptomics_st_rna_seq_workflow.png", 3652, cy),
    ("output/specific/metagenomics_shotgun_workflow.png", 5144, cy),
    ("output/specific/metagenomics_amplicon_workflow.png", 6612, cy),
    ("output/specific/metatranscriptomics_shotgun_workflow.png", 8084, cy),
]


# Bucle para dibujar el recuadro en cada análisis específico
for output_path, cx, cy in analisis:
    
    # Abrimos SIEMPRE la imagen original (importante)
    img = Image.open(input_path).copy()
    draw = ImageDraw.Draw(img)
    
    # Calcular esquinas desde el centro
    x0 = cx - width // 2
    y0 = cy - height // 2
    x1 = cx + width // 2
    y1 = cy + height // 2

    # Dibujar recuadro
    for i in range(grosor):
        draw.rectangle([x0-i, y0-i, x1+i, y1+i], outline=color)

    # Asegurar que el directorio existe
    os.makedirs(os.path.dirname(output_path), exist_ok=True)

    # Guardar
    img.save(output_path)
    print(f"✔ Guardado: {output_path}")
```
